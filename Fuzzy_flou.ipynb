```python
#Tp_exercoice
#les entreprises des logiciels informatique
import numpy as np

# Notes des candidats
echelle_notes = {
    "F":  (3, 5, 7),
    "MG": (5, 7, 9),
    "G":  (7, 9, 10),
    "VG": (9, 10, 10)
}

# Importance des critères
echelle_poids = {
    "M":  (0.3, 0.5, 0.7),
    "MH": (0.5, 0.7, 0.9),
    "H":  (0.7, 0.9, 1),
    "VH": (0.9, 1, 1)
}

# Afficher les résultats avec 4 décimales
np.set_printoptions(precision=4, suppress=True)
```

```python
notes = [
    [  # Candidat A1
        ["MG", "G",  "F",  "VG", "F"],  # DM1
        ["G",  "MG", "G",  "G",  "F"],  # DM2
        ["MG", "F",  "G",  "VG", "F"]   # DM3
    ],

    [  # Candidat A2
        ["G",  "VG", "VG", "VG", "VG"], # DM1
        ["G",  "VG", "VG", "VG", "MG"], # DM2
        ["MG", "VG", "G",  "VG", "G"]   # DM3
    ],

    [  # Candidat A3
        ["VG", "MG", "G",  "G",  "G"],  # DM1
        ["G",  "G",  "MG", "VG", "G"],  # DM2
        ["F",  "VG", "VG", "MG", "MG"]  # DM3
    ]
]
```

```python
notes_floues = np.array([
    [
        [echelle_notes[note] for note in ligne]
        for ligne in candidat
    ]
    for candidat in notes
], dtype=float)

# Afficher les évaluations de A1
print(notes_floues[0])
```

```python
X = notes_floues.mean(axis=1) #la moyens des 3 décideurs

for i in range(3):
    print(f"\nCandidat A{i + 1} — lignes C1 à C5 :")
    print(X[i])
```

```python
print(X[0, 0])
```

```python
# Troisième composante de chaque triplet
bornes_hautes = X[:, :, 2]

# Maximum par critère parmi les trois candidats
maximums = bornes_hautes.max(axis=0)

print("Maximums pour C1 à C5 :")
print(maximums)

# Normaliser chaque colonne séparément
R = np.zeros_like(X)

for j in range(5):
    R[:, j, :] = X[:, j, :] / maximums[j]

for i in range(3):
    print(f"\nMatrice normalisée de A{i + 1} :")
    print(R[i])
```

```python
poids = [
    ["H",  "VH", "VH", "VH", "M"],   # DM1
    ["VH", "VH", "H",  "VH", "MH"],  # DM2
    ["MH", "VH", "H",  "VH", "MH"]   # DM3
]

poids_flous = np.array([
    [echelle_poids[p] for p in ligne]
    for ligne in poids
], dtype=float)

# Moyenne des trois décideurs
W = poids_flous.mean(axis=0)

for j in range(5):
    print(f"Poids de C{j + 1} : {W[j]}")
```

```python
# 7. Question 4 : calculer la matrice normalisée pondérée on multiple chaque triplet normalisé pondéree par son critére 
```

```python
V = np.zeros_like(R)

for i in range(3):
    for j in range(5):
        V[i, j] = R[i, j] * W[j]

for i in range(3):
    print(f"\nMatrice pondérée de A{i + 1} :")
    print(V[i])
```

```python
ideal_positif = np.array([1, 1, 1])
ideal_negatif = np.array([0, 0, 0])

def distance_floue(triplet_1, triplet_2):
    difference = triplet_1 - triplet_2
    return np.sqrt(np.sum(difference ** 2) / 3)
```

```python
D_plus = np.zeros(3)
D_moins = np.zeros(3)

for i in range(3):
    for j in range(5):
        D_plus[i] += distance_floue(
            V[i, j], ideal_positif
        )

        D_moins[i] += distance_floue(
            V[i, j], ideal_negatif
        )

print("Distances à l'idéal positif :", D_plus)
print("Distances à l'idéal négatif :", D_moins)
```

```python
CC = D_moins / (D_moins + D_plus)

for i in range(3):
    print(f"A{i + 1} : CC = {CC[i]:.4f}")

# Le signe moins permet de classer du plus grand au plus petit
ordre = np.argsort(-CC)

print("\nClassement :")
for rang, indice in enumerate(ordre, start=1):
    print(f"{rang}. A{indice + 1} — score {CC[indice]:.4f}")
```
